# Data Understanding — Dirty Cafe Sales

## 1. Memuat Dataset

Dataset Dirty Cafe Sales merupakan data transaksi penjualan kafe yang berisi informasi mengenai ID transaksi, produk, jumlah pembelian, harga, total pengeluaran, metode pembayaran, lokasi transaksi, dan tanggal transaksi.

Tahap Data Understanding dilakukan untuk memahami struktur, karakteristik, dan kualitas dataset sebelum memasuki tahap persiapan data.

```python
import pandas as pd

df = pd.read_csv("../data/dirty_cafe_sales.csv")

df.head()
```

## 2. Memeriksa Ukuran Dataset

Pemeriksaan ukuran dataset dilakukan untuk mengetahui jumlah baris dan kolom. Informasi ini memberikan gambaran mengenai banyaknya data transaksi dan atribut yang tersedia untuk dianalisis.

```python
print("Jumlah baris dan kolom:", df.shape)
```

## 3. Memeriksa Nama Kolom

Pemeriksaan nama kolom bertujuan untuk mengetahui atribut yang tersedia dalam dataset serta memahami informasi yang direpresentasikan oleh setiap kolom.

```python
print("Nama kolom:")
print(df.columns.tolist())
```

## 4. Memeriksa Struktur dan Tipe Data

Pemeriksaan struktur dataset dilakukan untuk mengetahui tipe data dan jumlah nilai yang tersedia pada setiap kolom. Tahap ini membantu mengidentifikasi kolom yang memiliki nilai kosong serta kolom yang tipe datanya perlu diperiksa lebih lanjut.

```python
df.info()
```

## 5. Melihat Sampel Data

Sampel data ditampilkan untuk memahami bentuk data, isi setiap kolom, serta menemukan indikasi awal masalah kualitas data, seperti nilai kosong atau penanda `UNKNOWN` dan `ERROR`.

```python
df.head(10)
```

## 6. Memeriksa Data Duplikat

Pemeriksaan data duplikat dilakukan untuk mengetahui apakah terdapat baris yang memiliki seluruh nilai sama. Data duplikat perlu diidentifikasi karena dapat memengaruhi hasil perhitungan dan analisis apabila merepresentasikan transaksi yang tercatat lebih dari sekali.

```python
print("Jumlah baris duplikat:", df.duplicated().sum())
```

## 7. Memeriksa Nilai Unik

Pemeriksaan nilai unik dilakukan untuk mengetahui variasi nilai pada setiap kolom. Hasilnya dapat membantu mengidentifikasi kategori yang tersedia serta nilai tidak valid atau tidak konsisten, seperti `UNKNOWN` dan `ERROR`.

```python
for kolom in df.columns:
    print(f"\n--- {kolom} ---")
    print(df[kolom].value_counts(dropna=False).head(10))
```

## 8. Mengidentifikasi Masalah Kualitas Data

Pemeriksaan kualitas data dilakukan untuk menghitung jumlah nilai `UNKNOWN`, `ERROR`, dan nilai kosong pada setiap kolom. Hasil pemeriksaan digunakan untuk mengidentifikasi masalah awal yang perlu dipertimbangkan pada tahap Data Preparation.

```python
for kolom in df.columns:
    nilai = df[kolom].astype("string").str.strip().str.upper()

    unknown = nilai.eq("UNKNOWN").sum()
    error = nilai.eq("ERROR").sum()
    kosong = df[kolom].isna().sum()

    print(
        f"{kolom}: UNKNOWN={unknown}, "
        f"ERROR={error}, Kosong={kosong}"
    )
```

## 9. Kesimpulan Data Understanding

Berdasarkan pemeriksaan awal, dataset Dirty Cafe Sales terdiri dari 10.000 baris dan 8 kolom. Pemeriksaan duplikat dilakukan terhadap seluruh kolom. Dataset juga perlu diperiksa lebih lanjut karena terdapat nilai kosong serta penanda `UNKNOWN` dan `ERROR` pada beberapa kolom.

Selain itu, tipe data seluruh kolom perlu ditinjau sebelum analisis. Temuan ini menjadi dasar untuk melanjutkan proses ke tahap Data Preparation.

```python
print("Jumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])
print("Jumlah baris duplikat:", df.duplicated().sum())
print("\nJumlah nilai kosong per kolom:")
print(df.isna().sum())
```
