# Import library dan membaca dataset

```python
import pandas as pd

df = pd.read_csv("../data/dirty_cafe_sales.csv")

display(df.head())
```

# Memahami ukuran dan struktur dataset

```python
print("Jumlah baris dan kolom:", df.shape)

print("\nNama kolom:")
print(df.columns.tolist())

print("\nInformasi dataset:")
df.info()
```

# Memeriksa kualitas data

```python
print("Missing values:")
display(df.isnull().sum())

print("Jumlah baris duplikat:", df.duplicated().sum())

print("Jumlah nilai unik:")
display(df.nunique())
```

# Memeriksa nilai yang sering muncul

```python
for kolom in df.columns:
    print(f"\n--- {kolom} ---")
    display(df[kolom].value_counts(dropna=False).head(10))
```
